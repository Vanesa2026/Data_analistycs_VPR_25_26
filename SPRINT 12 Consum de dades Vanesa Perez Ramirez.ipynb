## Ejercicio 1
1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

Llista de publicacions (/posts)
Llista d’usuaris (/users)
Llista de tasques (/todos)

2. Mostra per pantalla:

La quantitat total de cada recurs.
El codi d'estat de cada petició.

```python
import requests
# Creamos variables de url y endpoints que utilizaremos para hacer la solicitud
url='https://jsonplaceholder.typicode.com/'
endpoints=['posts','users','todos']
# Como son tres endpoints diferentes a los que vamos a dar las mismas indicaciones, creamos un bucle for:
for endpoint in endpoints:
    solicitud_endpoint= requests.get(url+endpoint)
    print(f'El código de estado de la solicitud de {endpoint} es: {solicitud_endpoint.status_code}.')
    solicitud_endpoint_json=solicitud_endpoint.json()
    print(f'El número de registros de {endpoint} es {len(solicitud_endpoint_json)}\n') 

```

3. Fes una petició a una publicació inexistent per obtenir un error 404 i mostra el codi d'estat rebut.

```python
inexistente=requests.get(url+'posts/9999999')# Para que salga error 404 lo que hemos de hacer es buscar un dato en la lista de publicaciones que no exista:
if inexistente.status_code==200:
    print('La consulta se ha gestionado correctamente. Su código de estado es:', inexistente.status_code)
else:     
    print(f'El código de estatus es:{inexistente.status_code}. Es un error: {inexistente.reason}')
```

4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un user Id.
* Mostra la resposta JSON.
* Mostra el codi d'estat.

```python
# Datos de la nueva publicación ficticia

publicacion={'title':'Publicación ficticia con post','body':'Este es un ejercicio de clase en el que hago una publicación y este es el cuerpo del mensaje', 'userId':1}
# Hacemos la petición
peticion_post=requests.post(url+'posts',json=publicacion)
# Mostramos lo que nos solicitan
print(f'Esta es la respuesta en json:\n {peticion_post.json()}')

print(f'Este es el código de estado: {peticion_post.status_code}')

```

5. Fes una petició PATCH per modificar parcialment una publicació existent.

* Mostra la resposta JSON.
* Mostra el codi d'estat.

```python
# Creamos la modificación

actualizar_datos={'title':'El título nuevo es este', 'body':'El cuerpo nuevo será este'}

# Creamos la solicitud de modificación

peticion_patch=requests.patch(url+'posts/1',json=actualizar_datos)

print(f'Este es el código de estado: {peticion_patch.status_code}')

# Mostramos lo que nos piden:
print(f'Esta es la respuesta json:\n {peticion_patch.json()}')


```

6. Fes una petició DELETE sobre una publicació.

* Mostra la resposta JSON.
* Mostra el codi d'estat.

```python


# Creamos la solicitud de eliminación

peticion_eliminar=requests.delete(url+'posts/1')

# Mostramos lo que nos piden:
print(f'Esta es la respuesta json:\n {peticion_eliminar.json()}')

print(f'Este es el código de estado: {peticion_eliminar.status_code}')
```

```python
url='https://api.nobelprize.org/2.1/'
endpoints=['laureates','nobelPrizes']

peticion_api_real=requests.get(url+'laureates')

print(f'El código de estado de la petición es: {peticion_api_real.status_code}\n')

json_peticion_api_real=peticion_api_real.json()


# Tras revisar el json decido sacar los siguientes datos: 'id','knownName', 'nobelPrizes':{'awardYear':,'category':{'en'},'motivation:{'en'}}, de los tres primeros 'id' que aparecen
# Para ello preparo: 
# Tres primeros datos: 
primeros_id_json_peticion_api_real=json_peticion_api_real["laureates"][:3]

# De esos hago un bucle for con lo que se va a imprimir en bonito:

for i, premio in enumerate(primeros_id_json_peticion_api_real,start=1):
    # Datos que vamos a seleccionar:
    ganador_id=premio.get('id')
    nombre=premio.get('knownName',{}).get('en')

    for nobel in premio.get('nobelPrizes',[]):
        año_premio=nobel.get('awardYear')
        categoria=nobel.get('category',{}).get('en')
        motivo=nobel.get('motivation',{}).get('en')

        print(f'El ganador del premio Nobel del año {año_premio} en la categoría {categoria} fue {nombre} y el motivo de fue {motivo}. Su id en la base de datos es {ganador_id}.')
    print()



```
