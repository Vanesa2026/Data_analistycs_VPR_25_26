```python
df_trans_country=pd.merge(df_companies[['country','company_id']], df_transactions[['business_id','id']], left_on='company_id',right_on='business_id')[['country','id']]

df_products_country=pd.merge(df_trans_country[['country','id']],df_products_transactions, left_on='id',right_on='transaction_id')[['product_id','country']]

df_top5_sales=df_products_country['product_id'].value_counts().reset_index().sort_values('count',ascending=False).head(3)
top5_lista=list(df_top5_sales['product_id'])
print(top5_lista)

df_products_top5_country=df_products_country.loc[(df_products_country['product_id'].isin(top5_lista))]
df_products_top5_country
```

```python


df_company_products=pd.crosstab(df_products_top5_country['country'],df_products_top5_country['product_id'], normalize=True)

df_company_products
```

```python

df_company_products.plot.bar(colormap='Paired')
plt.xlabel('')
```

```python

df_company_products.plot.barh(stacked=True,colormap='Paired', figsize=(10,6))
plt.xlabel('')
```
