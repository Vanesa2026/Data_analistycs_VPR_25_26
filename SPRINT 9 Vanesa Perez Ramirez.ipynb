### Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. 

Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.

(EXTRA): Quina és la llargada mitjana de les paraules del text que has escrit? “Hola com va?” hauria de retornar (4+3+2) / 3 = 3


```python

import string
#había hecho la limpieza del texto dentro pero al ver que voy a utlizar el código limpieza más adelante lo he creado como una función separada

def limpiar_palabras(texto):
    """Dado un texto limpia símbolos y números"""
    
     #preparo que es lo que quiero quitar del texto. Está en la librería string, la tienen creada. He visto que no tiene incluido el símbolo¡ con lo que a la limpieza del texto se lo añado.
    limpiar_texto= string.punctuation + string.digits + '¡'

    #preparo la tabla de limpieza en la que le digo que no me va sustituir nada, pero que sí va a eliminar los símbolos y números que hay en la lista de 
    # limpieza_texto, esto es lo que llaman tabla de mapeo va a indicar a translate como se ha de modificar el string
    tabla_limpieza = str.maketrans("","",limpiar_texto)

    #ahora aplico la limpieza del texto usando el método translate al que le damos como argumento el mapa de mapeo que hemos creado para este caso
    texto_limpio =texto.translate(tabla_limpieza)
    #devuelve texto sin números ni símbolos
    return texto_limpio
#aquí he creado el contador de palabras
def contar_palabras(texto):
    """Dado un texto ordena alfabéticamente el texto, cuenta las palabras del texto y calcula la longitud media de las palabras. Devuelve una tupla con los dos valores"""
    texto_limpio=limpiar_palabras(texto)
    lista_palabras=texto_limpio.split()
    lista_palabras.sort()

    contador_palabras={}
   
    for palabra in lista_palabras:
        contador_palabras[palabra]=contador_palabras.get(palabra,0)+1

  

#esta es la parte extra nos solicitan lo he dejado dentro pero lo que hago que devuelva la función como valor para poder 
# reutilizar es el cáculo de la longitud media de las palabras
    sum=0
    for palabra in lista_palabras:
    #poner algo que quite todo y solo devuelva letras 
        sum= sum + len(palabra)
        llargada_mitjana = round(sum/len(lista_palabras),2)
    
    return contador_palabras,llargada_mitjana
```

# Diccionari invers (amb possibilitat de duplicats)

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.

Fixa’t en el següents exemples per veure l’output esperat.


```python
# Valores repetidos en un diccionario: 
def deteccion_valores_repetidos(diccionario_entrada):
    """Dado un diccionario proporciona los valores repetidos"""
    valores_con_repeticion=[]
    lista_valores=[]
    
    #esto lo hago para saber los valores repetidos.MIRAR QUE COÑO PASA PORQUE CLAVE NO LO DETECTA, AUNQUE NO LO NECESITO
    for clave,valor in diccionario_entrada.items():
        
        if valor in lista_valores:
            valores_con_repeticion.append(valor)
        else:
            lista_valores.append(valor)
    return valores_con_repeticion

def diccionario_reverse(diccionario):
    """Dado un diccionario realiza el reverse del diccionario, si el diccionario tiene valores iniciales que se duplican al hacer el reversea visa de la repetición
      y asocia una lista con los valores repetidos"""

   
    lista_valores_repetidos= deteccion_valores_repetidos(diccionario)
    reverse_diccionario={}
    reverse_diccionario_con_error={}

    if len(lista_valores_repetidos)==0:
        reverse_diccionario ={valor:clave for clave,valor in diccionario.items()}
        return reverse_diccionario
    else:
        for clave,valor in diccionario.items():
            if valor in lista_valores_repetidos:
                reverse_diccionario_con_error.setdefault(valor,[]).append(clave)
            else:
                reverse_diccionario_con_error.setdefault(valor,clave)
        for clave,valor in reverse_diccionario_con_error.items():
            if type(valor) is list:
                print(f"Hay claves con más de un valor asociados, la clave {clave} tiene los siguientes valores asociados {valor}")
        return reverse_diccionario_con_error   

```

[ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]


```python

def separador_datos_float(lista):
    """Recibe una lista y selecciona separa los elementos que se pueden transformar a float y los que no"""
   
    lista_datos_limpia=[]
    lista_flotants=[]
    lista_no_flotants=[]
    for item in lista:
        if isinstance(item,(list,tuple)):
            for subitem in item:
                lista_datos_limpia.append(subitem)
        else:
            lista_datos_limpia.append(item)

    for elemento in lista_datos_limpia:
        if type(elemento) is float or type(elemento) is int or validacion_float(elemento): 
            elemento=float(elemento)
            lista_flotants.append(elemento)
        else:
            lista_no_flotants.append(elemento)
    
    return [lista_no_flotants,lista_flotants]
```

Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:
    - longitud (int): Longitud de la contrasenya
    - majuscules (bool = True): Si hi ha d’aparèixer majúscules
    - minuscules (bool = True): Si hi ha d’aparèixer minúscules
    - numeros (bool = True): Si hi ha d’aparèixer números
    - signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)

Així doncs, si executem la funció de la següent manera:

    - Crear_contrasenya(10, True, True, True, True)
    - Hauríem d’obtenir un output (que hem de poder guardar) de l’estil:
        9Er,5Vn8P$

Assegura’t que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.   

(EXTRA) Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l’ordinador (com si l’haguéssim seleccionat i fet ctrl+copy).





```python


from numpy import random
import numpy as np
import string
import pyperclip as clip



def crear_contrasenya(largura, mayusc,minusc,num,simbolos):


    #como los boolean son 0 y 1 se pueden sumar con lo que tendremos el valor del denominador del size de los random
    denominador=mayusc+minusc+num+simbolos
    # La lógica del size de los random es que dividiendo con // aunque hay resto dá un número entero. 
    # El problema es que el resto son caracteres que hemos de calcular igualmente con lo que por un lado hago el cálculo del entero por el otro el cálculo del resto. 
    # Tomo la decisión de que el resto se lo aplicaré a los números, si números es false a las mayúsculas y si mayúsc es false a las minusculas, el resto nunca llegará a los símbolos
  
    #Calculo de la longitud del array en función de la longitud de la contraseña
    # En función del número de parámetros true el denominador es calculado, 
    # el número máximo de resto será cuando los cuatros parámetros estén en True con lo que el código está calculado para esta situación

    #Creo la lista de generadores que se han creado para poder poder concatenarlos al final y crear la matriz
    lista_activos=[]
 
    #genradores de valores aleatorios por variables:
    #generado de números:
    if num:
        generador_numeros=random.randint(largura,size=(((largura//denominador)+(largura%denominador))))
        lista_activos.append(generador_numeros)

    #generador de minúsculas
    #Para generar letras minúsculas
    if minusc:
        lista_minusculas=list(string.ascii_lowercase)
        generador_letras_min=random.choice(lista_minusculas, size=((largura//denominador)))
        lista_activos.append(generador_letras_min)
    #generador de mayusculas
    #Para generar mayúsculas:
    if mayusc:
        lista_mayusculas=list(string.ascii_uppercase)
        generador_letras_mayusc=random.choice(lista_mayusculas, size=((largura//denominador)))
        lista_activos.append(generador_letras_mayusc)
    #generador de símbolos:
    if simbolos:
        lista_simbolos=list(string.punctuation)
        generador_letras_simbolos=random.choice(lista_simbolos,size=((largura//denominador)))
        lista_activos.append(generador_letras_simbolos)

    # El resto de números hasta la lingitud deseada de la contraseña los reparto de la siguiente manera:
    if not num:
        if minusc:
            generador_letras_min_resto=random.choice(lista_minusculas, size=((largura%denominador)))
            lista_activos.append(generador_letras_min_resto)
            if mayusc:
                generador_letras_mayusc_resto=random.choice(lista_mayusculas, size=((largura%denominador)))
                lista_activos.append(generador_letras_mayusc_resto)

    #Creo la matriz con los valores que me ha dado cada función activa:
      
    matriz_contraseña=np.concatenate((lista_activos))
    #creo matriz permutada
   
    matriz_contraseña_permutada=random.permutation(matriz_contraseña)
    #paso la matriz a lista
    lista_contraseña=matriz_contraseña_permutada.tolist()
    #paso la lista a contraseña 
    contraseña="".join(lista_contraseña)
    contraseña_copiada=clip.copy(contraseña)
    return contraseña

```

Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. Necessita que processem les dades de manera automàtica, per extreure’n els resultats que necessita.

Utilitza el fitxer "historic_partits.txt"

Necessita un programa que retorni:

•	El nombre total de gols que ha fet cada equip.
•	El nom de l’equip més golejador.
•	El nom de l’equip més golejat
•	La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts)



```python
import re
import pandas as pd


def clasificacion_equipos(ruta):
    """De una ruta dada tras validarla, importa el archivo, lo lee y lo manipula para proporciona el nombre del máximo goleador, el nombrede goles más encajados 
    y la tabla de clasificaciones con los datos estadísticos de todos los equipos"""
    # como al hacer pruebas me he equivocado de archivo y rompía, para solucionar la carga de archivo erróneo el programa he añadido el siguiente try/except que te hace volver a cargar la ruta hasta que el archivo sea el correcto 

    while True:
#     # se le ha dado la opción al usuario de salir de programa durante la validación de la ruta y de que el archivo es txt. Devuelve None
        ruta_valida=validar_ruta_archivo(ruta)
        if ruta_valida==None:
            return None
        try:
            # Uso la función de validación definida en el ejercicio 3 del nivel 1
            with open(ruta_valida,'r',encoding='utf-8') as archivo_txt:
                    texto=archivo_txt.read()
            
            # A continuación se ha de limpiar el texto para poder trabajar con él. En este caso en el texto cuando se hace la lectura por líneas hay saltos de linea, tabuladores y guines entre los números.
            #Me daba un error de fuera de rango en el i del for i range(len(equipos_casa)he visto que era el len de equipos casa después de investigar he deducido que había una doble 
            # tabulación con que para arreglar esto he puesto primero el replace de la doble tabulación.
            texto=texto.strip().replace("\t\t","")

            # Limpio la lista de tabuladores salto de líneas y del guión que separa los números, lo separo todo por comas y finalmente hago el split con el separador de coma
            tabla_limpieza = str.maketrans("\t\n-\r",",,,,")
            texto_partidos_limpia =texto.translate(tabla_limpieza).strip().split(",")
        # Paso a números los string de goles
            for i in range(len(texto_partidos_limpia)):
                if texto_partidos_limpia[i].isnumeric():
                    texto_partidos_limpia[i]=int(texto_partidos_limpia[i])

            # Separo en listas por grupos de datos
            equipos_casa=texto_partidos_limpia[::4]
            goles_marcados=texto_partidos_limpia[1::4]
            goles_encajados=texto_partidos_limpia[2::4]
            equipos_visitante=texto_partidos_limpia[3::4]
        # Generación de diccionado de partidos con los resultados por equipo_casa, porque todos los equipos juegan tanto en casa como fuera de casa. 
        # En este caso hago un para la lista equipos_casa hago un set para quitar los equipos repetidos porque juegan muchos partidos en casa y si no hago el set se reptirá muchas veces 
        # la misma acción. 
            listado_equipos=set(equipos_casa)
        # Uso una comprehesion como generador:
            diccionario_partidos={equipo:{"goles encajados":0,"goles marcados":0,"partidos perdidos":0,"partidos ganados":0,"partidos empatados":0,"puntuación clasificación":0,} for equipo in listado_equipos}

        # Para saber los goles encajados vamos a trabajar con índices con lo que lo que haga será i in range(len(equipos_casa)), todas las listas tienen el valor correspondiente en la 
        # misma posición de cada lista creada:        
            for i in range(len(equipos_casa)):
                # Ahora aquí vamos a hacer un lógica encontrando valores en las listas:
                if goles_marcados[i]<goles_encajados[i]:#cuando los de casa pierden
                    diccionario_partidos[equipos_casa[i]]["partidos perdidos"] +=1
                    diccionario_partidos[equipos_visitante[i]]["partidos ganados"] +=1
                    diccionario_partidos[equipos_visitante[i]]["puntuación clasificación"] +=3
                    diccionario_partidos[equipos_casa[i]]["goles encajados"]+=goles_encajados[i]
                    diccionario_partidos[equipos_casa[i]]["goles marcados"]+=goles_marcados[i]
                    diccionario_partidos[equipos_visitante[i]]["goles encajados"] +=goles_marcados[i]
                    diccionario_partidos[equipos_visitante[i]]["goles marcados"]+=goles_encajados[i]
                if goles_marcados[i]== goles_encajados[i]:#hay empate
                    diccionario_partidos[equipos_casa[i]]["partidos empatados"] +=1
                    diccionario_partidos[equipos_casa[i]]["puntuación clasificación"] +=1
                    diccionario_partidos[equipos_visitante[i]]["partidos empatados"] +=1
                    diccionario_partidos[equipos_visitante[i]]["puntuación clasificación"] +=1
                    diccionario_partidos[equipos_casa[i]]["goles encajados"]+=goles_encajados[i]
                    diccionario_partidos[equipos_casa[i]]["goles marcados"]+=goles_marcados[i]
                    diccionario_partidos[equipos_visitante[i]]["goles encajados"] +=goles_marcados[i]
                    diccionario_partidos[equipos_visitante[i]]["goles marcados"]+=goles_encajados[i]
                if goles_marcados[i]> goles_encajados[i]: #los de casa ganan 
                    diccionario_partidos[equipos_casa[i]]["partidos ganados"] +=1
                    diccionario_partidos[equipos_casa[i]]["puntuación clasificación"] +=3
                    diccionario_partidos[equipos_visitante[i]]["partidos perdidos"] +=1
                    diccionario_partidos[equipos_casa[i]]["goles encajados"]+=goles_encajados[i]
                    diccionario_partidos[equipos_casa[i]]["goles marcados"]+=goles_marcados[i]
                    diccionario_partidos[equipos_visitante[i]]["goles encajados"] +=goles_marcados[i]
                    diccionario_partidos[equipos_visitante[i]]["goles marcados"]+=goles_encajados[i]            
            
        # Para poder el diccionario ordenado por valor, lo he buscado y se ha de hacer a través de una función o un lambda o sorted(), estudiado y aprendido y queda de la siguiente manera:
        # #Número de goles que ha hecho cada equipo:
        # calsificación global, con goles marcados, encajados, partidos ganados, partidos perdidos, y ordenado por clasificación:
            diccionario_partidos_ordenado=dict(sorted(diccionario_partidos.items(), key= lambda item: item[1]["puntuación clasificación"], reverse=True))
            listado_clasificacion=[]
            for item in diccionario_partidos_ordenado:
                listado_clasificacion.append(item)
            clasificacion_enumerada =list(enumerate(tuple(listado_clasificacion),start=1))
    # de la tabla global extraigo los goles marcados por equipo:
            goles_marcados_por_equipo=[(clave,diccionario_partidos_ordenado[clave]["goles marcados"]) for clave,valor in diccionario_partidos_ordenado.items()]
            # Para que se vea como una tabla pasamos el diccionario a un dataframe indicando que las claves sean el índice:
            tabla_clasificaciones= pd.DataFrame.from_dict(diccionario_partidos_ordenado, orient='index')
        # Responder a la pregunta: equipos máximo goleador. Se ha de hacer con Lambda o función:
            max_goleador= max(diccionario_partidos,key= lambda equipo: diccionario_partidos[equipo]["goles marcados"])
        # Responder minimo goleador
            max_encajados= max(diccionario_partidos, key=lambda equipo: diccionario_partidos[equipo]["goles encajados"])
       
            print(f"El equipo que ha marcado más goles es {max_goleador}.\nEl equipo que más goles ha encajado es {max_encajados}.")
            return goles_marcados_por_equipo, clasificacion_enumerada ,max_goleador , max_encajados , tabla_clasificaciones
            #Aquí pongo el except para que si el archivo no es el correcto en lugar de romperse el programa que solicite de nuevo la ruta
        except:
            # print("el archivo no corresponde a la función")
            # break
            nueva_ruta = input("El archivo que has indicado no corresponde al que se necesita para esta función. Por favor, indica la ruta correcta o 'Salir' si quieres finalizar:")
            if nueva_ruta.lower().strip()=="salir":
                return None
            ruta=nueva_ruta
```
